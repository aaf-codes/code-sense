```python
from pypdf import PdfReader

pdf_path = "GEN AI Syllabus.pdf"

reader = PdfReader(pdf_path)

text = ""

for page in reader.pages:
    page_text = page.extract_text()

    if page_text:
        text += page_text + "\n"

print("PDF loaded successfully!")
print("Pages:", len(reader.pages))
print("Characters:", len(text))
```

```python
chunk_size = 500
chunk_overlap = 50

chunks = []

start = 0

while start < len(text):

    end = start + chunk_size
    chunk = text[start:end]

    if chunk.strip():
        chunks.append(chunk)

    start += chunk_size - chunk_overlap

print("Number of chunks:", len(chunks))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded!")
```

```python
document_embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True
)

print("Document embeddings created!")
print("Embedding shape:", document_embeddings.shape)
```

```python
import faiss

faiss.normalize_L2(document_embeddings)

dimension = document_embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(document_embeddings)

print("FAISS index created successfully!")
print("Number of stored vectors:", index.ntotal)
```

```python
def retrieve_relevant_chunks(question, top_k=3):

    # Convert the question into an embedding
    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    )

    # Normalize question embedding
    faiss.normalize_L2(question_embedding)

    # Search FAISS
    scores, indices = index.search(
        question_embedding,
        top_k
    )

    results = []

    for i, score in zip(indices[0], scores[0]):

        if i != -1:
            results.append({
                "text": chunks[i],
                "score": float(score)
            })

    return results
```

```python
question = "What is the objective of this course?"

results = retrieve_relevant_chunks(question)

for i, result in enumerate(results):

    print(f"\n--- Retrieved Chunk {i+1} ---")
    print(f"Similarity Score: {result['score']:.4f}")
    print(result["text"])
```

```python
response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "user",
            "content": "Explain Generative AI in one sentence."
        }
    ]
)

print(response["message"]["content"])
```

```python
def ask_question(question, top_k=5):

    results = retrieve_relevant_chunks(question, top_k)

    context = "\n\n".join(result["text"] for result in results)

    prompt = f"""
You are an academic assistant.

Answer the question using ONLY the information provided
in the context below.

Give a clear and concise answer.
If the question asks for different types, list each type
and briefly explain it using the available context.

Do not make up information.

If the answer truly cannot be found in the context, say:
"The answer is not available in the document."

Context:
--------------------
{context}
--------------------

Question:
{question}

Answer:
"""

   

    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response["message"]["content"], results
```

```python
question = input("Enter your question: ")

answer, retrieved_chunks = ask_question(question)

print("\n" + "=" * 60)
print("GENERATED ANSWER")
print("=" * 60)
print(answer)
```

```python
# In this we are building a RAG (Retrieval-Augmented Generation) system 
# that answers questions from an academic PDF.
# PDF → Chunks → Embeddings → FAISS → Relevant Information → Llama 3.2 → Answer.
```
